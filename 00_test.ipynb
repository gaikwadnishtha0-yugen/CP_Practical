```python
#If elif else statement for marks of the student
#Above 95 - Grade A, 85-94 - Grade C, 75-84 - Grade D, Below 75 - No grade
marks = int(input("Enter your marks: "))
if marks > 95:
    print("Grade A")
elif marks >= 85:
    print("Grade C")
elif marks >= 75:
    print("Grade D")
else:
    print("No grade")
```
